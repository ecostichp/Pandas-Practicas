```python
from ipychart import Chart

# 1. Definir los datos para las dos series
datos = {
    'labels': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo'],
    'datasets': [
        {
            'label': 'Ventas 2025',
            'data':[1,2,3],  # Datos de la primera serie
            'backgroundColor': 'rgba(54, 162, 235, 0.5)',   # Azul transparente
            'borderColor': 'rgba(54, 162, 235, 1)',       # Azul sólido
            'borderWidth': 1
        },
        {
            'label': 'Ventas 2026',
            'data':[3,2,1],  # Datos de la segunda serie
            'backgroundColor': 'rgba(255, 99, 132, 0.5)',   # Rojo transparente
            'borderColor': 'rgba(255, 99, 132, 1)',       # Rojo sólido
            'borderWidth': 1
        }
    ]
}

# 2. Configurar el título y comportamiento del gráfico
opciones = {
    'plugins': {
        'title': {
            'display': True,
            'text': 'Comparativa de Ventas Mensuales (2025 vs 2026)',
            'font': {'size': 16}
        },
        'legend': {
            'position': 'top'  # Muestra la leyenda de las dos series arriba
        }
    }
}

# 3. Crear y mostrar el gráfico de barras agrupadas
chart = Chart(data=datos, kind='bar', options=opciones)
chart
```

```python
# Se puede guardar después de mostrarla, sin problemas
chart.to_image("./mi_grafica_ventas.jpg")
```
