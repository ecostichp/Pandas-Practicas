```python
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# 1. Crear datos de ejemplo con Pandas
np.random.seed(42)
df = pd.DataFrame(
    {
        'Categoria': np.random.choice(['A', 'B', 'C'], size=100),
        'ValorX': np.random.randn(100),
        'ValorY': np.random.randn(100) * 5 + 10,
    }
)

# 2. Configurar estilo de Seaborn
sns.set_theme(style='whitegrid')

# 3. Crear el archivo PDF multipágina
with PdfPages('reporte_datos.pdf') as pdf:
    # --- Gráfica 1: Scatter plot con Seaborn ---
    fig1, ax1 = plt.subplots(figsize=(8, 5))
    sns.scatterplot(
        data=df, x='ValorX', y='ValorY', hue='Categoria', ax=ax1, palette='deep'
    )
    ax1.set_title('Relación entre ValorX y ValorY')
    plt.tight_layout()
    pdf.savefig(fig1)  # Guardar página 1
    plt.close(fig1)

    # --- Gráfica 2: Histograma con Seaborn ---
    fig2, ax2 = plt.subplots(figsize=(8, 5))
    sns.histplot(data=df, x='ValorY', hue='Categoria', kde=True, ax=ax2)
    ax2.set_title('Distribución de ValorY')
    plt.tight_layout()
    pdf.savefig(fig2)  # Guardar página 2
    plt.close(fig2)

    # --- Gráfica 3: Tabla resumen generada con Pandas y Matplotlib ---
    fig3, ax3 = plt.subplots(figsize=(8, 4))
    ax3.axis('tight')
    ax3.axis('off')
    resumen = df.groupby('Categoria')[['ValorX', 'ValorY']].mean().reset_index()
    tabla = ax3.table(
        cellText=resumen.values,
        colLabels=resumen.columns,
        loc='center',
        cellLoc='center',
    )
    tabla.scale(1, 1.5)
    ax3.set_title('Resumen Promedio por Categoría', fontsize=12, pad=15)
    pdf.savefig(fig3, bbox_inches='tight')  # Guardar página 3
    plt.close(fig3)

print("¡Reporte PDF generado exitosamente!")
```
